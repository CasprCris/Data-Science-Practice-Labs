We can replace the missing values of the column <code>'bedrooms'</code> with the mean of the column  <code>'bedrooms' </code> using the method <code>replace()</code>. Don't forget to set the <code>inplace</code> parameter to <code>True</code>


```python
mean=df['bedrooms'].mean() #calculates the mean of the column "bedrooms" (why not take the Median here?) There is one resident with 33 bedrooms!
df['bedrooms'].replace(np.nan,mean, inplace=True) #replaces the NaN values of the column "bedrooms" with the mean
```

We also replace the missing values of the column <code>'bathrooms'</code> with the mean of the column  <code>'bathrooms' </code> using the method <code>replace()</code>. Don't forget to set the <code> inplace </code>  parameter top <code> True </code>


```python
mean=df['bathrooms'].mean() #calculate the mean of the column "bathrooms" (maximum is 8 here so its fine?)
df['bathrooms'].replace(np.nan,mean, inplace=True) #replace all NaN values with the mean
```

```python
#check if there are any NaN values left
print("number of NaN values for the column bedrooms :", df['bedrooms'].isnull().sum())
print("number of NaN values for the column bathrooms :", df['bathrooms'].isnull().sum())
```

# Module 5: Model Evaluation and Refinement


Import the necessary modules:


```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
print("done")
```

We will split the data into training and testing sets:


```python
#create a list of features / column names
features =["floors", "waterfront","lat" ,"bedrooms" ,"sqft_basement" ,"view" ,"bathrooms","sqft_living15","sqft_above","grade","sqft_living"]    
X = df[features] #assign features to X
Y = df['price'] #assign Target variable to Y

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.15, random_state=1) #Split data in Test and Train Set with 15% used for testing

print("number of test samples:", x_test.shape[0]) #.shape returns Rows/columns. [0] accesses  the first element
print("number of training samples:",x_train.shape[0]) #.shape returns Rows/columns. [0] accesses  the first element
```

### Question 9

Create and fit a Ridge regression object using the training data, set the regularization parameter to 0.1, and calculate the R^2 using the test data. Take a screenshot of your code and the value of the R^2. You will need to submit it for the final project.


```python
from sklearn.linear_model import Ridge
```

```python
RR = Ridge(alpha = 0.1) #Creating a RidgeRegression Object with hyperparameter alpha set to 0.1
RR.fit(x_train, y_train) #Train the model using the training data
RR.score(x_test, y_test) #Calculate the R^2 using the test data
```

### Question 10

Perform a second order polynomial transform on both the training data and testing data. Create and fit a Ridge regression object using the training data, set the regularisation parameter to 0.1, and calculate the R^2 utilising the test data provided. Take a screenshot of your code and the R^2. You will need to submit it for the final project.


```python
pr = PolynomialFeatures(degree=2) #Create a second order Polynomial Feature Object
x_train_pr = pr.fit_transform(x_train) #Polynomial Feature Transformation of the x Train data
x_test_pr = pr.fit_transform(x_test) #Polynomial Feature Transformation of the x test data
RR = Ridge(alpha = 0.1) #Create a Ridge Regression Object with regularisation parameter set to 0.1
RR.fit(x_train_pr, y_train) #Train the model
RR.score(x_test_pr, y_test) #Calculate the test score
```
